```python
from radon.raw import analyze
from radon.complexity import cc_visit
from radon.metrics import mi_visit

file_path = "refactored_code/chatgpt/Rich_decimal.py"

with open(file_path, "r", encoding="utf-8") as f:
    code = f.read()

# Calculate metrics
raw = analyze(code)
loc = raw.loc

complexity = cc_visit(code)
cc = complexity[0].complexity if complexity else 0

mi = mi_visit(code, multi=True)

print("=" * 40)
print("ChatGPT Refactored Metrics")
print("=" * 40)
print(f"Lines of Code (LOC): {loc}")
print(f"Cyclomatic Complexity (CC): {cc}")
print(f"Maintainability Index (MI): {mi:.2f}")
```

```python
import pandas as pd

results = pd.DataFrame({
    "Function": ["Rich_decimal"],
    "Model": ["ChatGPT"],
    "LOC": [loc],
    "Cyclomatic Complexity": [cc],
    "Maintainability Index": [round(mi, 2)]
})

results
```

```python
results.to_csv(
    "results/chatgpt_results.csv",
    index=False
)

print("Results saved successfully!")
```

```python
from radon.raw import analyze
from radon.complexity import cc_visit
from radon.metrics import mi_visit


def analyze_python_file(file_path, label):
    with open(file_path, "r", encoding="utf-8") as f:
        code = f.read()

    raw = analyze(code)
    complexity = cc_visit(code)
    mi = mi_visit(code, multi=True)

    print("=" * 50)
    print(label)
    print("=" * 50)
    print(f"Lines of Code (LOC): {raw.loc}")
    print(f"Cyclomatic Complexity: {sum(c.complexity for c in complexity)}")
    print(f"Maintainability Index: {mi:.2f}")
    print()
```
